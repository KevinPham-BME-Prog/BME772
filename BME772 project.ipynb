import dataframes

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('condensed performance scores - ML testing.csv', header = None, delim_whitespace= False)
print (df.head())
```

import the feature extraction values (found by using matlab)

```python
import numpy as np
```

```python
psdalpha = df[3][1:]
psdbeta = df[4][1:]
psddelta = df[5][1:]
psdtheta = df[6][1:]
hacta = df[7][1:]
hactb = df[8][1:]
hactd = df[9][1:]
hactt = df[10][1:]
moba = df[11][1:]
mobb = df[12][1:]
mobd = df[13][1:]
mobt = df[14][1:]
compa = df[15][1:]
compb = df[16][1:]
compd = df[17][1:]
compt = df[18][1:]
performance = df[19][1:]
```

obtain experience level (label)

```python
experience = df[20][1:]
```

combine features into stack

```python
#feactures = np.stack([psdalpha, psdbeta, psddelta, psdtheta, hacta, hactb, hactd, hactt,moba,mobb, mobd, mobt, compa, compb, compd, compt], axis = 1)
feactures = np.stack([ psdbeta, psdtheta, hacta, hactb,moba,mobb, mobd, compa, compb, compd, compt], axis = 1)
```

various data features were found to contribute to the overall ML algorithm
these values was psdbeta, psdtheta, hacta, hactb,moba,mobb, mobd, compa, compb, compd, compt

set up random forest classifier

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
#imports all important libraries
```

```python
valuetrain, valuetest, labeltrain, labeltest = train_test_split(feactures, experience, test_size = 0.3, random_state = 42)
#makes a 70 - 30 split inorder to create a testing and validation split
```

make the random forest classifier

```python
classif = RandomForestClassifier(n_estimators= 9, max_depth = 3, random_state = 0)
classif.fit(valuetrain,labeltrain)
```

```python
print(valuetest)
```

```python
y_pred = classif.predict(valuetest)
acc = accuracy_score(labeltest, y_pred)
print('The accuracy is %f%%' %(acc*100))
```

return feature importance (note this is the finalized version as it was ran through multiple times to examine what values was being considered to set up the random forest algorithm)

```python
importances = classif.feature_importances_
for i,v in enumerate(importances):
    print('Feature: %0d, Score: %.5f' % (i,v))
```

```python
from sklearn.model_selection import cross_val_score
from sklearn import svm
scores = cross_val_score(classif, feactures, experience, cv =6)
print(scores)
print("%0.2f accuracy with a standard deviation of %0.2f" % (scores.mean(), scores.std()))
```
